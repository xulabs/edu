## Introduction to Diffusion Models for Medical Image Generation
**Level: Beginner**

Diffusion models are a class of generative AI models that learn to create new images
by reversing a gradual noising process. Instead of predicting a label (classification)
or a pixel-level mask (segmentation), a diffusion model learns to generate entirely new,
realistic images from random noise.

This notebook builds a small DDPM (Denoising Diffusion Probabilistic Model) trained on
a single MedNIST class, so you can see the full generative pipeline: adding noise,
training a model to remove it, and finally generating brand-new synthetic medical
images from pure noise.
